Delivery_Time_Model

```python
import pandas as pd 
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_excel('delivery_time.csv.xlsx')
df.head(5)
```

```python
X = df[['Delivery Time']]
y= df[['Sorting Time']]
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
X1 = 14
y=(model.coef_*X1)+model.intercept_
y
```

```python
X2 = 20
y=(model.coef_*X2)+model.intercept_
y
```

```python
X3 = 26
y=(model.coef_*X3)+model.intercept_
y
```
