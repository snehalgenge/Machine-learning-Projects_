Salary_Data_Model

```python
import pandas as pd   
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("Salary_Data.csv.xlsx")
df.head(5)
```

```python
X = df[['YearsExperience']]
y = df[['Salary']]
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
X1 = 1.4
y = (model.coef_*X1)+model.intercept_
y
```

```python
X2 = 1.7
y = (model.coef_*X2)+model.intercept_
y
```

```python
X3 = 3.5
y = (model.coef_*X3)+model.intercept_
y
```
